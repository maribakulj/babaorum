# Babaorum lecteur — R6 : extraction des lectures (CPU, aucun calcul de modèle)

```
# Extraction légère des sorties R6 (CPU) : stem, truth, hyp par ligne, sans treillis
import glob, gzip, json, os, hashlib
os.makedirs('/kaggle/working/r6_hyp', exist_ok=True)
for p in sorted(glob.glob('/kaggle/input/**/B?_dev.jsonl.gz', recursive=True)):
    arm = os.path.basename(p)[:2]
    out = f'/kaggle/working/r6_hyp/{arm}_hyp.jsonl.gz'
    n = 0
    with gzip.open(p, 'rt') as f, gzip.open(out, 'wt') as g:
        for line in f:
            r = json.loads(line)
            g.write(json.dumps({'stem': r['stem'], 'truth': r['truth'], 'hyp': r['hyp']}, ensure_ascii=False) + '\n')
            n += 1
    print(arm, p, n, 'lignes', os.path.getsize(out), 'octets', hashlib.sha256(open(out, 'rb').read()).hexdigest())
```
